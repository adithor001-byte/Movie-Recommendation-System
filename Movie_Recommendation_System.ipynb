```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
tfidf = TfidfVectorizer(
    stop_words='english',
    max_features=50000
)
```

```python
tfidf_matrix = tfidf.fit_transform(movies_model['tags'])
```

```python
tfidf_matrix.shape
```

```python
movie_indices = pd.Series(
    movies_model.index,
    index=movies_model['title']
)

movie_indices = movie_indices[~movie_indices.index.duplicated(keep='first')]
```

```python
def recommend_movies(title, n=10):

    if title not in movie_indices:
        return "Movie not found"

    idx = movie_indices[title]

    scores = cosine_similarity(
        tfidf_matrix[idx],
        tfidf_matrix
    ).flatten()

    top_indices = scores.argsort()[-n-1:][::-1]

    top_indices = [
        i for i in top_indices
        if i != idx
    ][:n]

    return movies_model.iloc[top_indices][
        ['title', 'vote_average', 'genres_text']
    ]
```

```python
from scipy.sparse import csr_matrix

user_ids = ratings_model['userId'].unique()
movie_ids = ratings_model['movieId'].unique()

user_to_idx = {user_id: i for i, user_id in enumerate(user_ids)}
movie_to_idx = {movie_id: i for i, movie_id in enumerate(movie_ids)}

print("Users:", len(user_to_idx))
print("Movies:", len(movie_to_idx))
```

```python
rows = ratings_model['userId'].map(user_to_idx)
cols = ratings_model['movieId'].map(movie_to_idx)

user_item_matrix = csr_matrix(
    (
        ratings_model['rating'].values,
        (rows, cols)
    ),
    shape=(len(user_ids), len(movie_ids))
)

print("Matrix shape:", user_item_matrix.shape)
print("Stored ratings:", user_item_matrix.nnz)
```

```python
from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(
    n_components=50,
    random_state=42
)

user_factors = svd.fit_transform(user_item_matrix)

print("Training complete!")
print("User factors shape:", user_factors.shape)
```

```python
movie_factors = svd.components_.T

print("Movie factors shape:", movie_factors.shape)
```

```python
def recommend_collaborative(user_id, n=10):

    # Check whether user exists
    if user_id not in user_to_idx:
        return "User not found"

    # Get user's learned preference vector
    user_index = user_to_idx[user_id]
    user_vector = user_factors[user_index]

    # Compare user preferences with every movie
    scores = movie_factors @ user_vector

    # Get highest-scoring movies
    top_indices = scores.argsort()[::-1]

    # Movies already rated by this user
    rated_movies = set(
        ratings_model[
            ratings_model['userId'] == user_id
        ]['movieId']
    )

    recommendations = []

    for idx in top_indices:

        movie_id = movie_ids[idx]

        # Don't recommend movies the user already rated
        if movie_id not in rated_movies:
            recommendations.append(movie_id)

        if len(recommendations) == n:
            break

    # Convert movie IDs to movie information
    result = movies_model[
        movies_model['id'].isin(recommendations)
    ][['id', 'title', 'vote_average', 'genres_text']].copy()

    # Keep recommendation order
    order = {movie_id: i for i, movie_id in enumerate(recommendations)}

    result['order'] = result['id'].map(order)

    result = result.sort_values('order')

    return result.drop(columns='order')
```

```python
def hybrid_recommend(movie_title, user_id, n=10):

    # ---------- CONTENT SCORE ----------
    if movie_title not in movie_indices:
        return "Movie not found"

    if user_id not in user_to_idx:
        return "User not found"

    movie_idx = movie_indices[movie_title]

    content_scores = cosine_similarity(
        tfidf_matrix[movie_idx],
        tfidf_matrix
    ).flatten()

    # ---------- COLLABORATIVE SCORE ----------
    user_idx = user_to_idx[user_id]
    user_vector = user_factors[user_idx]

    collaborative_scores = movie_factors @ user_vector

    # ---------- NORMALIZE SCORES ----------
    content_scores = (
        content_scores - content_scores.min()
    ) / (
        content_scores.max() - content_scores.min()
    )

    collaborative_scores = (
        collaborative_scores - collaborative_scores.min()
    ) / (
        collaborative_scores.max() - collaborative_scores.min()
    )

    # ---------- COMBINE ----------
    hybrid_scores = (
        0.6 * content_scores +
        0.4 * collaborative_scores
    )

    # ---------- REMOVE ORIGINAL MOVIE ----------
    hybrid_scores[movie_idx] = -1

    # ---------- TOP MOVIES ----------
    top_indices = hybrid_scores.argsort()[::-1]

    rated_movies = set(
        ratings_model[
            ratings_model['userId'] == user_id
        ]['movieId']
    )

    recommendations = []

    for idx in top_indices:

        movie_id = movie_ids[idx] if idx < len(movie_ids) else None

        if movie_id is not None and movie_id not in rated_movies:
            recommendations.append(idx)

        if len(recommendations) == n:
            break

    return movies_model.iloc[recommendations][
        ['title', 'vote_average', 'genres_text']
    ]
```

```python
# Map movie ID -> collaborative movie index
collab_movie_to_idx = {
    movie_id: i
    for i, movie_id in enumerate(movie_ids)
}

# For every movie in movies_model, store its collaborative index
movies_model['collab_idx'] = movies_model['id'].map(
    collab_movie_to_idx
)

print("Movies with collaborative data:",
      movies_model['collab_idx'].notna().sum())
```

```python
def hybrid_recommend(movie_title, user_id, n=10):

    if movie_title not in movie_indices:
        return "Movie not found"

    if user_id not in user_to_idx:
        return "User not found"

    # Content score
    movie_idx = movie_indices[movie_title]

    content_scores = cosine_similarity(
        tfidf_matrix[movie_idx],
        tfidf_matrix
    ).flatten()

    # Collaborative score
    user_idx = user_to_idx[user_id]
    user_vector = user_factors[user_idx]

    collab_scores = movie_factors @ user_vector

    # Create full-length collaborative score
    full_collab_scores = np.zeros(len(movies_model))

    for i, collab_idx in enumerate(movies_model['collab_idx']):
        if pd.notna(collab_idx):
            full_collab_scores[i] = collab_scores[int(collab_idx)]

    # Normalize both scores
    content_scores = (
        content_scores - content_scores.min()
    ) / (
        content_scores.max() - content_scores.min()
    )

    full_collab_scores = (
        full_collab_scores - full_collab_scores.min()
    ) / (
        full_collab_scores.max() - full_collab_scores.min()
    )

    # Hybrid score
    hybrid_scores = (
        0.6 * content_scores +
        0.4 * full_collab_scores
    )

    # Don't recommend the movie itself
    hybrid_scores[movie_idx] = -1

    # Movies already rated by user
    rated_movies = set(
        ratings_model[
            ratings_model['userId'] == user_id
        ]['movieId']
    )

    # Rank
    top_indices = hybrid_scores.argsort()[::-1]

    recommendations = []

    for idx in top_indices:

        movie_id = movies_model.iloc[idx]['id']

        if movie_id not in rated_movies:
            recommendations.append(idx)

        if len(recommendations) == n:
            break

    return movies_model.iloc[recommendations][
        ['id', 'title', 'vote_average', 'genres_text']
    ]
```

```python
from sklearn.model_selection import train_test_split

ratings_eval = ratings_model.sample(
    n=100000,
    random_state=42
)

train_eval, test_eval = train_test_split(
    ratings_eval,
    test_size=0.2,
    random_state=42
)

print("Training ratings:", len(train_eval))
print("Testing ratings:", len(test_eval))
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

global_mean = train_eval['rating'].mean()

baseline_predictions = np.full(
    len(test_eval),
    global_mean
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        test_eval['rating'],
        baseline_predictions
    )
)

print("Baseline RMSE:", baseline_rmse)
```

```python
from scipy.sparse import csr_matrix
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics import mean_squared_error
import numpy as np

# Create mappings from the evaluation training data
eval_user_ids = train_eval['userId'].unique()
eval_movie_ids = train_eval['movieId'].unique()

eval_user_to_idx = {
    user_id: i
    for i, user_id in enumerate(eval_user_ids)
}

eval_movie_to_idx = {
    movie_id: i
    for i, movie_id in enumerate(eval_movie_ids)
}

# Convert training ratings to sparse matrix
train_rows = train_eval['userId'].map(eval_user_to_idx)
train_cols = train_eval['movieId'].map(eval_movie_to_idx)

eval_matrix = csr_matrix(
    (
        train_eval['rating'].values,
        (train_rows, train_cols)
    ),
    shape=(
        len(eval_user_ids),
        len(eval_movie_ids)
    )
)

# Train SVD
eval_svd = TruncatedSVD(
    n_components=20,
    random_state=42
)

eval_user_factors = eval_svd.fit_transform(eval_matrix)
eval_movie_factors = eval_svd.components_.T

print("Evaluation model trained!")
print("Users:", len(eval_user_ids))
print("Movies:", len(eval_movie_ids))
```

```python
predictions = []
actual = []

for _, row in test_eval.iterrows():

    user_id = row['userId']
    movie_id = row['movieId']

    # Only evaluate users/movies seen during training
    if (
        user_id in eval_user_to_idx
        and movie_id in eval_movie_to_idx
    ):

        user_idx = eval_user_to_idx[user_id]
        movie_idx = eval_movie_to_idx[movie_id]

        prediction = (
            eval_user_factors[user_idx]
            @ eval_movie_factors[movie_idx]
        )

        predictions.append(prediction)
        actual.append(row['rating'])

model_rmse = np.sqrt(
    mean_squared_error(actual, predictions)
)

print("SVD RMSE:", model_rmse)
print("Test ratings evaluated:", len(actual))
```

```python
import gradio as gr

def movie_recommender_ui(movie_title, user_id):

    try:
        user_id = int(user_id)
    except:
        return "Please enter a valid User ID."

    result = hybrid_recommend(
        movie_title,
        user_id,
        n=10
    )

    if isinstance(result, str):
        return result

    output = ""

    for i, row in result.reset_index(drop=True).iterrows():
        output += (
            f"{i+1}. {row['title']}  "
            f"(Rating: {row['vote_average']})\n"
            f"   Genres: {row['genres_text']}\n\n"
        )

    return output


demo = gr.Interface(
    fn=movie_recommender_ui,
    inputs=[
        gr.Textbox(
            label="Movie",
            placeholder="Enter a movie title"
        ),
        gr.Number(
            label="User ID",
            value=1
        )
    ],
    outputs=gr.Textbox(
        label="Recommended Movies"
    ),
    title="🎬 Movie Recommendation System",
    description="Hybrid recommender using Content-Based Filtering + Collaborative Filtering"
)

demo.launch()
```
